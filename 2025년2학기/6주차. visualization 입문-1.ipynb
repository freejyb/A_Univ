## 데이터 시각화(data visualization)

### 개괄

- 데이터를 그래프, 차트, 지도, 플롯 등 다양한 형태로 시각적으로 표현하여 데이터를 쉽게 이해하고 분석할 수 있도록 도와주는 기법입니다. 
- 데이터 시각화는 복잡한 데이터의 패턴, 추세, 관계 등을 명확하게 드러내어 의사결정 과정에서 중요한 역할을 합니다.

- 데이터 분석 결과를 쉽게 이해할 수 있도록 시각적으로 표현하고 전달되는 과정
- 데이터 시각화의 목적은 도표라는 수단을 통해 정보를 명확하고 효과적으로 전달하는 것

### 데이터 시각화의 중요성

- 1. 직관적인 이해: 숫자나 표로만 구성된 데이터를 그래프나 차트로 변환함으로써 데이터의 흐름이나 관계를 직관적으로 이해할 수 있습니다.
- 2. 패턴과 추세 발견: 시각화를 통해 데이터에서 숨겨진 패턴, 추세, 이상치를 쉽게 발견할 수 있습니다.
- 3. 의사결정 지원: 기업의 경영진이나 연구자들이 빠르고 정확하게 결정을 내리는 데 필요한 인사이트를 제공합니다.
- 4. 효과적인 커뮤니케이션
- : 데이터를 시각적으로 표현하면 이해하기 쉬워, 다른 사람들과의 커뮤니케이션이 효율적으로 이루어집니다.

### 주요 데이터 시각화 유형

- 1. 막대 차트 (Bar Chart):
- 카테고리별 데이터를 비교할 때 유용합니다. (예시) 경찰청 별 범죄 검거건수
- 2. 선 그래프 (Line Graph):
- 시간에 따른 데이터 변화를 나타낼 때 사용합니다.
- 예: 연도별 범죄 현황 등.

- 3. 파이 차트 (Pie Chart): 전체 중에서 각 부분이 차지하는 비율을 나타낼 때 사용합니다.
- 예: 시장 점유율, 예산 분배 등.
- 4. 산점도 (Scatter Plot): 두 변수 간의 관계를 나타낼 때 유용합니다.
- 예: 키와 몸무게의 관계, 광고 비용과 판매량의 관계 등.
- 5. 히스토그램 (Histogram): 데이터의 분포를 나타내는 데 사용하며, 특정 값의 빈도를 시각화합니다.
- 예: 연령대별 인구수 등.
- 6. 히트맵 (Heatmap): 데이터의 밀도나 빈도를 색으로 표현합니다.
- 예: 웹사이트의 클릭 맵, 상관관계 매트릭스 등.
- 7. 박스 플롯 (Box Plot): 데이터의 중앙값, 사분위수, 이상치를 시각적으로 표현하여 데이터의 분포를 파악합니다.
- 예: 시험 점수의 분포, 연봉의 분포 등.
- 8. 지도 시각화 (Geographical Map Visualization): 데이터가 지리적 위치와 연관될 때 사용합니다.
- 예: 인구 밀도 지도, 판매 지역별 매출 등.

#### 선그리기

```python
import matplotlib.pyplot as plt
import matplotlib as mpl

# 🔧 한글 폰트 설정
mpl.rcParams['font.family'] = 'Malgun Gothic'  # Windows용
mpl.rcParams['axes.unicode_minus'] = False     # 마이너스 깨짐 방지

# 데이터 입력
years = [2020, 2021, 2022, 2023, 2024]
crime_counts = [100, 200, 250, 220, 350]

# 그래프 설정
# figsize=(8, 5) 는 가로 8인치, 세로 5인치 크기로 그래프 크기를 지정합니다.
# 그래프의 전체적인 크기 조절에 사용됩니다.

plt.figure(figsize=(8, 5))
plt.plot(years, crime_counts, marker='o', linestyle='-', linewidth=2)

# 제목 및 축 레이블
# title() → 그래프 상단의 제목 설정
# xlabel() → X축(가로)의 이름 설정
# ylabel() → Y축(세로)의 이름 설정
# fontsize= → 각 텍스트의 글자 크기 지정

plt.title("연도별 범죄 건수 변화", fontsize=14)
plt.xlabel("연도", fontsize=12)
plt.ylabel("범죄 건수", fontsize=12)

# X축 눈금을 1년 간격으로 고정
plt.xticks(years)

# Y축 범위 조정
# Y축 값의 범위를 직접 지정합니다.
# 최소값: 0, 최대값: 데이터의 최대값보다 약간 큰 값 → 여유 공간 확보용
# 보기 좋은 그래프를 만들기 위해 활용

plt.ylim(0, max(crime_counts) + 50)

# 격자 추가
plt.grid(True)

# 그래프 출력
plt.show()
```

#### 범례 표시하기

```python
# 범례 (Legend) : 그래프에 데이터의 종류를 표시하기 위한 텍스트

# 범례 표시 방법 :
# 1) plot() 함수에 label 문자열을 지정
# 2) matplotlib.pyplot 모듈의 legend() 함수를 호출
```

```python
import matplotlib.pyplot as plt
import matplotlib as mpl

# 🔧 한글 폰트 설정
mpl.rcParams['font.family'] = 'Malgun Gothic'  # Windows용
mpl.rcParams['axes.unicode_minus'] = False     # 마이너스 깨짐 방지

# 데이터 입력
years = [2020, 2021, 2022, 2023, 2024]
crime_counts = [100, 200, 250, 220, 350]

# 그래프 설정
plt.figure(figsize=(8, 5))

# 선 그래프 + 범례 라벨 지정
plt.plot(years, crime_counts, marker='o', linestyle='-', linewidth=2, label='연도별 범죄 추이')

# 제목 및 축 레이블
plt.title("연도별 범죄 건수 변화", fontsize=14)
plt.xlabel("연도", fontsize=12)
plt.ylabel("범죄 건수", fontsize=12)

# X축 눈금을 1년 간격으로 고정
plt.xticks(years)

# Y축 범위 조정
plt.ylim(0, max(crime_counts) + 50)

# 격자 추가
plt.grid(True)

# ✅ 범례 표시
plt.legend(loc='best')  # 'best'는 자동으로 적절한 위치에 배치

# 그래프 출력
plt.show()
```

```python
# 범례위치 지정하기
```

```python
# upper right'	오른쪽 위 (기본값)
# 'upper left'	왼쪽 위
# 'lower left'	왼쪽 아래
# 'lower right'	오른쪽 아래
# 'right'	그래프 오른쪽 바깥쪽
# 'center left'	가운데 왼쪽
# 'center right'	가운데 오른쪽
# 'lower center'	아래 중앙
# 'upper center'	위쪽 중앙
# 'center'	정중앙
# 'best'	자동 배치 (충돌 없는 위치)
```

```python
locations = ['upper left', 'lower right', 'center', 'right']

for loc in locations:
    plt.figure(figsize=(6, 4))
    plt.plot(years, crime_counts, marker='o', label='연도별 범죄 추이')
    plt.title(f"범례 위치: {loc}")
    plt.legend(loc=loc)
    plt.grid(True)
    plt.xticks(years)
    plt.ylim(0, max(crime_counts) + 50)
    plt.show()
```

#### 선종류 선택하기 

```python
# 데이터를 표현하기 위해 그려지는 선의 종류를 지정하는 방법
# 선 종류를 나타내는 문자열 또는 튜플을 이용해서 다양한 선의 종류를 구현할 수 있습니다.
# Keyword: plt.plot(), 포맷 문자열, 선 종류, linestyle, Solid, Dashed, Dotted, Dash-dot
```

```python
# Line style (선 스타일) : 문자 형태로 지정가능
# - : solid line
# -- : dashed line
# -. : dash-dot line
# : : dotted line
```

```python
import matplotlib.pyplot as plt

# 데이터
years = [2020, 2021, 2022, 2023, 2024]
crime_counts = [100, 200, 250, 220, 350]

# 위치와 선 스타일을 짝지어 리스트 구성
locations = ['upper left', 'lower right', 'center', 'right']
linestyles = ['-', '--', '-.', ':']  # 실선, 점선, 혼합선, 점점점선

# 반복 출력
for loc, ls in zip(locations, linestyles):
    plt.figure(figsize=(6, 4))
    plt.plot(
        years,
        crime_counts,
        linestyle=ls,
        marker='o',
        label=f'연도별 범죄 추이 ({ls})'
    )
    plt.title(f"범례 위치: {loc}")
    plt.legend(loc=loc)
    plt.grid(True)
    plt.xticks(years)
    plt.ylim(0, max(crime_counts) + 50)
    plt.show()
```

#### Matplotlib 색상 지정하기

```python
# 기본색상 : r, g, b 

# Color (색깔) : 약자 또는 컬러이름 또는 ‘#RGB’로 지정가능
# b : blue
# g : green
# r : red
# y : yellow
# k : black
```

```python
import matplotlib.pyplot as plt

# 데이터
years = [2020, 2021, 2022, 2023, 2024]
crime_counts = [100, 200, 250, 220, 350]

# 설정값: 범례 위치, 선 스타일, 선 색상
locations = ['upper left', 'lower right', 'center', 'right']
linestyles = ['-', '--', '-.', ':']
colors = ['blue', 'green', 'red', 'purple']  # 각 선의 색상

# 반복 출력
for loc, ls, color in zip(locations, linestyles, colors):
    plt.figure(figsize=(6, 4))
    plt.plot(
        years,
        crime_counts,
        linestyle=ls,
        color=color,
        marker='o',
        label=f'연도별 범죄 추이 ({ls}, {color})'
    )
    plt.title(f"범례 위치: {loc}")
    plt.legend(loc=loc)
    plt.grid(True)
    plt.xticks(years)
    plt.ylim(0, max(crime_counts) + 50)
    plt.show()
```

```python
# Easy Stacked Charts with Matplotlib and Pandas

### https://matplotlib.org/stable/gallery/lines_bars_and_markers/bar_stacked.html
```

```python
# x축에 Year, y축에 value를 보여주는 bar chart를 그리되, 월별로 bar segment를 stack해서 보려는 경우.
```

```python
import pandas as pd

data = [[2020, 2020, 2020, 2021, 2021, 2021, 2022, 2022, 2022],
        ['Jan', 'Feb', 'Mar', 'Jan', 'Feb', 'Mar', 'Jan', 'Feb', 'Mar'],
        [1, 2, 3, 4, 5, 6, 7, 8, 9]]

rows = zip(data[0], data[1], data[2])

headers = ['Year', 'Month', 'Value']

df = pd.DataFrame(list(rows), columns=headers)

df
```

```python
# 테이블 피벗하기
# ['Jan','Feb','Mar']] 부분은 월별로 순서를 정렬하게 위해 적용

pivot_df = df.pivot(index='Year', columns='Month', values='Value')
pivot_df = pivot_df[['Jan','Feb','Mar']].copy()
pivot_df
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

# Stacked Bar Chart
pivot_df.plot.bar(stacked=True, figsize=(10,7))
```

```python
# 범례(legend)의 순서를 Bar segment와 일치시키기 위해 legend='reverse'를 추가--> 범례 순서 변경
pivot_df.plot.bar(stacked=True, legend='reverse', figsize=(10,7))
```

```python
# 넘파이, 판다스 . 플로트리 라이브러리 불러오기
import numpy as np
import pandas as pd
import plotly.offline as pyo

# 샘플 데이터 생성
df = pd.DataFrame(np.random.randn(100,2), columns=['물가', '소비'])
```

```python
# 그래프 그리기
# 첫 번째 선 그래프 한 줄의 데이터를 변수에 저장
# 두 번째 선 그래프 한 줄의 데이터를 변수에 저장
# 모든 그래프의 데이터를 하나의 리스트에 저장
# 이 데이터를 사용해서 완성된 그래프를 그린다.
pyo.iplot([{'x':df.index, 'y':df['물가'], 'name': '물가'}, {'x':df.index, 'y':df['소비'], 'name': '소비'}])
```

```python
import numpy as np
import pandas as pd

# 라이브러리 불러오기
import plotly.offline as pyo
import plotly.graph_objs as go

df = pd.DataFrame(np.random.randn(100,2), columns=['A', 'B'])

# 라인 그래프 A, 라인 그래프 B의 데이터를 각각 할당합니다.
trace1 = go.Scatter(x = df.index, y = df['A'], name = 'A')
trace2 = go.Scatter(x = df.index, y = df['B'], name = 'B')

# 라인 그래프 A, B를 하나의 리스트에 담습니다.
# 최종적으로 그래프를 그리는데 사용될 데이터입니다.
data = [trace1, trace2]

# 이 데이터를 사용해서 그래프를 그립니다.
pyo.iplot(data)
```
