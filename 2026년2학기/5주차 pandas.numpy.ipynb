- 파이썬 라이브러리(Python Library)?

- 파이썬 프로그래밍 언어에서 특정 기능을 수행하기 위해 미리 작성된 코드의 모음입니다.

- 표준 라이브러리(Standard Library): 파이썬 설치 시 기본적으로 제공되는 라이브러리입니다. 파일 입출력, 문자열 처리, 수학 계산, 네트워크 통신, 날짜 및 시간 처리 등 기본적인 기능들을 제공합니다. 예시로 os, sys, math, datetime 등이 있습니다.
- 서드파티 라이브러리(Third-Party Library): 표준 라이브러리에 포함되지 않은 기능을 추가로 제공하는 라이브러리로, 파이썬 커뮤니티나 기업이 개발하여 공개한 라이브러리입니다. 이 라이브러리들은 파이썬 패키지 인덱스(PyPI)에서 쉽게 설치할 수 있으며, 데이터 분석, 웹 개발, 머신러닝 등 다양한 분야에서 활용됩니다. 
- 예시: numpy, pandas, requests, scikit-learn 등이 있습니다.
- 라이브러리를 활용하면 복잡한 기능도 간단하게 구현할 수 있어, 개발의 효율성을 크게 높여줍니다.

```python
# 첫번째 방법
# 2차원 표에서 column 단위로 데이터를 표현하여 dataframe 생성
# ==> column명을 딕셔너리의 key로 (아래에서 crime) ,  데이터는 딕셔너리의 value로
```

```python
from pandas import DataFrame

crime={"살인":[10,5,20], "강도":[55,65,40], "절도":[150,200,350]}

police =["서울경찰청", "부산경찰청", "인천경찰청"]

df=DataFrame(crime, police)
print(df)
```

### Dictionary를 바탕으로 한번 DataFrame을 생성

```python
# DataFrame 생성하기

# Dictionary를 바탕으로 한번 DataFrame을 생성

# 딕셔너리(dict)를 통해 각 칼럼에 대한 데이터를 저장한 후
# 딕셔너리를 DataFrame 클래스의 생성자 인자로 넘겨주면 DataFrame 객체 생성     
```

```python
from datetime import datetime, timedelta
import pandas as pd
import random

# 랜덤 시드 설정
random.seed(42)

# 데이터 개수 설정
data_count = 100

# 날짜 범위 설정
start_date = datetime(2022, 1, 1)
end_date = datetime(2022, 12, 31)
date_range = [start_date + timedelta(days=random.randint(0, (end_date - start_date).days)) for _ in range(100)]

# ID, 나이, 성별, 소득, 범죄건수 랜덤 생성
id_list = [i+1 for i in range(data_count)]
age_list = [random.randint(18, 80) for _ in range(data_count)]
gender_list = ['남성', '여성']
gender = [random.choice(gender_list) for _ in range(data_count)]
income_list = [random.randint(20000, 150000) for _ in range(data_count)]
crime_count_list = [random.randint(0, 10) for _ in range(data_count)]

# 데이터프레임 생성
data = {
    
    '날짜': date_range,
    'ID' : id_list,
    '나이': age_list,
    '성별': gender,
    '소득': income_list,
    '범죄건수': crime_count_list
}

df = pd.DataFrame(data)

df

```

```python
# index 날짜로 변경 set_index()
df.set_index("날짜",inplace=True)
```

# NumPy

- NumPy : Numerical Python
- 다차원(n차원 배열 생성 : 1차원, 2차원..) 배열(array) 연산기능을 모아둔 패키지
- 다차원 배열 객체와 이를 효율적으로 다루기 위한 다양한 함수들을 제공합니다. 
- 넘파이는 빠르고 유연한 배열 연산을 통해 수학, 과학, 공학 계산 등에 많이 사용되며, 다른 데이터 분석 및 머신러닝 라이브러리(예: Pandas, SciPy, Scikit-learn 등)의 기반이 되기도 합니다.
- 벡터, 행렬, 다차원 배열을 위한 파이썬 객체 및 이 객체들을 이용한 수치연산, 통계 계산 등의 기능
- 행렬, 배열의 수학연산을 벡터연산(일종의 병렬 연산)을 하여 성능을 개선 (연산속도 측면)

## http://www.numpy.org/

## 넘파이의 주요 특징과 기능

- 1. N차원 배열 객체 (ndarray):
- 넘파이의 핵심 데이터 구조로, 동일한 데이터 타입의 값들이 담긴 다차원 배열입니다. 
- 리스트와 유사하지만, 배열 크기가 고정되어 있고, 수치 계산에 훨씬 효율적입니다.
- 2. 빠른 연산 속도:
- 넘파이는 C 언어로 구현되어 있어 파이썬의 기본 리스트보다 훨씬 빠르게 수학적 연산을 수행할 수 있습니다.
- 3. 벡터화 연산:
- 반복문 없이 배열 전체에 대한 연산을 한 번에 수행할 수 있습니다. 
- 이를 통해 코드가 간결해지고 연산 속도가 빨라집니다.
- 4. 브로드캐스팅(Broadcasting):
- 크기가 다른 배열 간의 연산을 자동으로 맞춰 수행할 수 있습니다. 
- 예를 들어, 배열과 스칼라 값 간의 연산에서 배열의 각 요소에 스칼라 값이 적용됩니다.
- 5. 다양한 수학 함수:
- 선형대수, 통계, 푸리에 변환 등의 다양한 수학 함수들을 제공합니다.
- 6. 난수 생성:
- 다양한 분포에서 난수를 생성할 수 있어, 시뮬레이션이나 머신러닝에서 많이 활용됩니다.
- 7.파일 입출력:
- 배열 데이터를 바이너리 형식이나 텍스트 형식(CSV, TSV 등)으로 저장하거나 불러올 수 있습니다.

### 배열이란?

```python
# 데이터 처리를 효율적으로 하기 위한 자료구조
# 배열 안에는 숫자나 문자열  모두를 담을 수 있다. 단, 배열 안에서는 동일 종류의 데이터만 담을 수 있다.
# - 리스트는 다른 유형의 데이터를 담을 수 있지만 배열은 불가
# 리스트 : [1,2,3] , [안산, 대학교, 경찰], [1,2, 안산, 4]
# 배열 : ([1,2,3]) , ([안산, 대학교, 경찰])
```

```python
# 배열의 종류(차원: Dimension ) : 1차원, 2차원, 3차원
# n차원의 배열 (ndarray)
# 1) 1차원 배열 : 범죄건수 ([61000,59000,56500]) -->  리스트와 유사하나 , 연산 방법의 차이로 속도가 빠르다.
# 2) 2차원 배열 : 서울범죄건수, 부산범죄건수, ( [61000,59000,56500], [ 200000, 210000, 2250000])
# 3) 3차원 배열 : 2차원 배열이 수개 모여 있는 형태 
```

### 배열 만들기

- 배열을 만들기 위한 함수
- 1) np.array  : 초기 값을 지정하여 배열 만들기
- 2) np.arrange : 순차적인 숫자 이용 배열 만들기
- 3) np. rand : 랜덤하게 숫자 배열 만들기

```python
### np.array
```

```python
# 만드는 방법
# np.array ([나열된 데이터 객체])
```

```python
# 넘파이 불러오기
import numpy as np
# np.array ([나열된 데이터 객체])
# stock1 리스트 데이터를 배열 형태로 변환

person=[15000, 14500, 16000, 17000]
arr1=np.array([person])
arr1
```

```python
# 데이터 타입(형식) 확인하기

type(arr1)
```

- 기본 내장함수 range 와 유사함 : 연속된 숫자로 배열 만들기
- 코드 방법 :  np.arrange([])

```python
import numpy as np
np.arange(10)
```

```python
# np.arange(시작번호, 끝번호)
np.arange(2, 10)
```

```python
# np.arange(시작번호, 끝번호, 건너뛰는 수)
np.arange(1, 30, 7)
```
